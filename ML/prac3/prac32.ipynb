```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([
    [1,100],
    [2,250],
    [3,350],
    [4,450],
    [5,550],
    [6,650],
    [7,750]
], dtype=float)

y = np.array([
    600,500,700,800,900,1000,1100
], dtype=float)

m = len(y)
```

```python
print(X)
```

```python
# Feature Scaling
X = (X - X.mean(axis=0)) / X.std(axis=0)

# Add bias column
X = np.c_[np.ones(X.shape[0]), X]

# Initialize weights
O = np.zeros(X.shape[1])

lr = 0.01
epochs = 1000
```

```python
print(np.isnan(X).any())
print(np.isnan(y).any())
print(np.isnan(O).any())
```

```python
cost_history = []

for _ in range(epochs):
    indices = np.random.permutation(m)

    for i in indices:

        xi = X[i]      
        yi = y[i]     
        y_pred = np.dot(xi, O)

        dO = (y_pred - yi) * xi

        # Update weights
        

    y_pred_all = X @ O
    cost = (1 / (2 * m)) * np.sum((y_pred_all - y) ** 2)
    cost_history.append(cost)

print("Weights:", O)
y_pred = X @ O
```

```python
print("Weights:", O)
```

```python
plt.figure(figsize=(7,5))
plt.plot(cost_history)
plt.title("Stochastic Gradient Descent - Cost vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Cost")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(7,5))

plt.plot(range(1, len(y)+1), y, 'bo-', label="Actual")
plt.plot(range(1, len(y)+1), y_pred, 'rs--', label="Predicted")

plt.xlabel("Sample")
plt.ylabel("Target Value")
plt.title("Multiple Linear Regression using SGD")
plt.legend()
plt.grid(True)
plt.show()
```

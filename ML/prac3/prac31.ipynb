```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.array([1,2,6,4,5,6,7],dtype = float);
y=np.array([10,9,8,7,6,5,4],dtype=float)
```

```python
o0=0
o1=0
n=len(x);
lr=0.01
epoch=1000
```

```python
for i in range(epoch):
    y_pred = o0+o1*x
    error = y_pred-y

    do1 = (1/n)*np.sum(error*x)
    do0 = (1/n)*np.sum(error)

    o0 = o0-lr*do0
    o1=o1-lr*do1
```

```python
print("Slope =",o1)
print("Intercept =",o0)
```

```python
plt.scatter(x,y,color='red')
plt.plot(x,o1*x+o0,color='blue')
plt.show()
```

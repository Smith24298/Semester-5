```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([
    [1,100],[2,250],[3,350],[4,450],[5,550],[6,650],[7,750]
],dtype = float)
y = np.array([
    600,500,700,800,900,1000,1100
],dtype=float)
m = len(y)
```

```python
print(X)
```

```python
X = (X - X.mean(axis=0)) / X.std(axis=0)
X =  np.c_[np.ones(X.shape[0]), X]
O = np.zeros(X.shape[1])
lr=0.01
epochs = 1000
```

```python
print(np.isnan(X).any())
print(np.isnan(y).any())
print(np.isnan(O).any())
```

```python
for _ in range(epochs):
    y_pred = np.dot(X, O)

    dO = (1 / m) * np.dot(X.T, (y_pred - y))

    O = O - lr * dO
```

```python
print("Weights:", O)
```

```python
plt.plot(range(1, len(y)+1), y, 'bo-', label="Actual")
plt.plot(range(1, len(y)+1), y_pred, 'rs--', label="Predicted")
plt.xlabel("Sample")
plt.ylabel("Target Value")
plt.title("Actual vs Predicted")
plt.legend()
plt.grid(True)
plt.show()
```

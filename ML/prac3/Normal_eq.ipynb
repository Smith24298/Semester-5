```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([
    [1,100],
    [2,250],
    [3,350],
    [4,450],
    [5,550],
    [6,650],
    [7,750]
], dtype=float)
y = np.array([
    600,500,700,800,900,1000,1100
], dtype=float)
```

```python
X = (X - X.mean(axis=0)) / X.std(axis=0)
X = np.c_[np.ones(X.shape[0]), X]
```

```python
print(X)
```

```python
theta = np.linalg.inv(X.T @ X) @ X.T @ y
```

```python
print(theta)
```

```python
y_pred = X @ theta
```

```python
print("\nPredictions:")
print(y_pred)
```

```python
plt.figure(figsize=(7,5))

plt.plot(range(1, len(y)+1), y, 'bo-', label="Actual")
plt.plot(range(1, len(y)+1), y_pred, 'r*-', label="Predicted")

plt.xlabel("Sample")
plt.ylabel("Target")
plt.title("Normal Equation: Actual vs Predicted")
plt.legend()
plt.grid(True)

plt.show()
```

```python
mse = np.mean((y - y_pred) ** 2)
rmse = np.sqrt(mse)

print("MSE :", mse)
print("RMSE:", rmse)
```

```python
from pdf2image import convert_from_path
import pytesseract
```

```python
pdf_path = "pdftest.pdf"
```

```python
pages = convert_from_path(pdf_path, dpi=300)
```

```python
for i,page in enumerate(pages):
    text = page_text = pytesseract.image_to_string(page)
    print(f"this is page {i}")
    print(text)
```
